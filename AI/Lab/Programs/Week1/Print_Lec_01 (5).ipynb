```python
#Search for a number x in this tuple using loop:   [1, 4, 9, 16, 25, 36, 49, 64, 81, 100]
nums = (1, 4, 9, 16, 25, 36, 30, 64, 30, 100)
x = 30  # Number to search

idx = 0
for el in nums:
    if(el == x):
        print("found at index", idx)
       # break
    idx += 1
```
