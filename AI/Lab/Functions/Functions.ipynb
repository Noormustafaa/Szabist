```python
# Write a Program to finde the Avrg of 3 numbers !

def cal_average(num1, num2, num3):
    total = num1 + num2 + num3
    average = total / 3
    return average

print(cal_average(10, 20, 30))
```

```python
# Types Of Functions 

# 1. Built-in Functions
#print(len("Hello, World!"))  # Built-in function to get the length of a string
print(len("Hello, World!"))  # Built-in function to get the length of a string

#type()  # Built-in function to get the type of an object
print(type(42))  # Built-in function to get the type of an object

#range
range(5)  # Built-in function to create a range of numbers
for i in range(5):
    print(i)  # Built-in function to create a range of numbers

    
```

```python
# user-defined Functions
#Default perameters. assign default values to parameters in the function definition. If the caller does not provide a value for that parameter, the default value will be used.

def cal_product(num1=3, num2=2):
    product = num1 * num2
    return product

cal_product()  # Uses default values (3 and 2)



```
