```python
#Dictionary in python is a collection of key-value pairs. Each key is unique and maps to a specific value. Dictionaries are defined using curly braces {} and can contain elements of different data types.
#ex
dict1 = {'name': 'Alice',
        
        'age': 25, 

        'city': 'New York',

        'Subjects': ['Math', 'Science', 'English'], # we can also store list in dictionary as value
        'marks'   : (85, 90, 78) # we can also store tuple in dictionary as value
        
         
        }
print(dict1)  # Output: {'name': 'Alice', 'age': 25, 'city': 'New York'}
print(type(dict1))  # Output: <class 'dict'>

# note in values we can assign any data type like list, tuple, set, dictionary etc. but in keys we can only assign immutable data types like string, integer, float, tuple etc. because keys are unique and hashable.


```

```python
# Properties of dictionary in python:
# 1 : it is unordered collection of key-value pairs.

# we can access the values in a dictionary using their corresponding keys. For example, to access the value associated with the key 'name', we can use dict1['name'], which will return 'Alice'.
#ex
dic1={
    'name' : 'Alice',
    'age' : 25,
    'Subjects':['Math', 'Science', 'English']


}

print(dic1['name'])  # Output: Alice
print(dic1['age'])  # Output: 25
print(dic1['Subjects'])  # Output: ['Math', 'Science', 'English']

dic1['age'] = 26  # Updating the value associated with the key 'age'
print(dic1['age'])  # Output: 26
```

```python
# we can also create an empty dictionary and add key-value pairs to it later. For example, we can create an empty dictionary using dict2 = {} and then add key-value pairs using the syntax dict2[key] = value.
dict2 = {}
print(dict2)  # Output: {}
dict2['name'] = 'samee'
print(dict2['name'])  # Output: {'name': 'samee'}
```

```python
# nasted Dictionary in python is a dictionary that contains another dictionary as a value. It allows us to create a hierarchical structure of data, where each key can have its own set of key-value pairs. We can access the values in a nested dictionary using multiple keys, separated by square brackets [].

#ex: 
students={

    'name':'Ali',

    'Subjects':{
        'math': 85,
        'science': 90,
        'English': 78
    }
}

#This is how we can create a nested dictionary in python. In this example, the key 'Subjects' contains another dictionary as its value, which has three key-value pairs representing the marks of the student in different subjects. We can access the marks of a specific subject using the syntax students['Subjects']['math'], which will return 85.
print(students['Subjects']['math'])  # Output: 85
print(students)
```

```python
# Dictionary Methods in python:
#mydic.keys() : This method returns a view object that displays a list of all the keys in the dictionary. The view object is dynamic, meaning that if the dictionary is modified, the view object will reflect those changes.
#ex:
dic3={
    'name': 'Alice',
    'age': 25,

    'Subjects':
    {
        'mathematics': 85,
        'science': 90,
    }


}

print(dic3.keys())  # Output: dict_keys(['name', 'age', 'Subjects'])

print(list(dic3.keys()))  # Output: ['name', 'age', 'Subjects'] # type casting the dict_keys object to a list using the list() function, which allows us to work with the keys as a standard Python list.

print(len(dic3))  # Output: 3 (the number of key-value pairs in the dictionary)
```

```python
# mydic.values() : This method returns a view object that displays a list of all the values in the dictionary. The view object is dynamic, meaning that if the dictionary is modified, the view object will reflect those changes.
#ex:
dic4={
    'name': 'Alice',
    'age': 25,

    'Subjects':
    {
        'mathematics': 85,
        'science': 90,
    }


}
print(dic4.values())  # Output: dict_values(['Alice', 25, {'mathematics': 85, 'science': 90}])
print(len(dic4))  # Output: 3 (the number of key-value pairs in the dictionary)
```

```python
# mydic.items() : This method returns a view object that displays a list of all the key-value pairs in the dictionary as tuples. The view object is dynamic, meaning that if the dictionary is modified, the view object will reflect those changes.
#ex
dic5={
    'name': 'Alice',
    'age': 25,
    'subjects': ['Math', 'Science', 'English']
}
print(dic5.items())  # Output: dict_items([('name', 'Alice'), ('age', 25), ('subjects', ['Math', 'Science', 'English'])])

pairs=list(dic5.items())  # Converting the dict_items object to a list of tuples
print(pairs[0])  # Output: ('name', 'Alice') (the first key-value pair as a tuple)
```

```python
# .get() method in python dictionary is used to retrieve the value associated with a specified key. It takes the key as an argument and returns the corresponding value if the key exists in the dictionary. If the key does not exist, it returns a default value (which can be specified as a second argument) or None if no default value is provided.
#ex:
dic6={
    'name': 'Alice',
    'age': 25,
    'city': 'New York'
}
print(dic6.get('name'))  # Output: Alice (the value associated with the key 'name')
print(dic6['name'])  # we can also write this 
```

```python
#update() method in python dictionary is used to update the values of existing keys or add new key-value pairs to the dictionary. It takes another dictionary or an iterable of key-value pairs as an argument and updates the original dictionary with the new values.
#ex:
dic7={
    'name': 'Alice',
    'age': 25,

}
dic7.update({'age': 26, 'city': 'New York'})  # Updating the value of 'age' and adding a new key-value pair for 'city'

print(dic7)  # Output: {'name': 'Alice', 'age': 26, 'city': 'New York'}

# or 
new_dict={'age':27,'city': 'Los Angeles'}
dic7.update(new_dict)  # Updating the value of 'age' and 'city' using another dictionary
print(dic7)  # Output: {'name': 'Alice', 'age': 27, 'city': 'Los Angeles'}
```
