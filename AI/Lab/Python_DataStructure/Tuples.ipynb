```python
#Tuples is the immutable data structure in python. It means that once a tuple is created, its elements cannot be modified, added, or removed. Tuples are defined using parentheses () and can contain elements of different data types.
```

```python
tuple1 = (1, 2, 3, 4, 5)
print(tuple1)  # Output: (1, 2, 3, 4, 5)
print(type(tuple1))  # Output: <class 'tuple'>
print(tuple1[0])  # Output: 1
print(tuple1[1:4])  # Output: (2, 3, 4)

#tup[0] = 10  # This will raise a TypeError because tuples are immutable
```

```python
# Empty tuple
empty_tuple = ()
print(empty_tuple)  # Output: ()
print(type(empty_tuple))  # Output: <class 'tuple'>

# if There is a Single element in the tuple then we have to add a comma after the element to indicate that it is a tuple.
tuple2 = (1,)
print(tuple2)  # Output: (1,)
print(type(tuple2))  # Output: <class 'tuple'>

# but if we don't add a comma after the single element, it will be considered as an integer and not a tuple.
tuple3 = (1)
print(tuple3)  # Output: 1
print(type(tuple3))  # Output: <class 'int'>

tuple4 = (90.1)
print(tuple4)  # Output: 90.1
print(type(tuple4))  # Output: <class 'float'>
```

```python
# tuples methods are limited because tuples are immutable. The only methods available for tuples are count() and index().
tuple5 = (1, 2, 3, 3, 2, 2, 3)
print(tuple5.count(2))  # Output: 3 (count of 3 in the tuple)
print(tuple5.index(3))  # Output: 2 (index of the first occurrence
```
