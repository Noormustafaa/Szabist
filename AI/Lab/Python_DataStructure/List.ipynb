```python
# List in python !

my_list = [1, 2, 3, 4, 5]

print(my_list)
print(type(my_list))

#Note In List we can access the elements by using index number and also we can change the value of any element by using index number.

print(my_list[0])  # Accessing the first element

my_list[0] = 10  # Changing the value of the first element
print(my_list)

print(len(my_list))  # Length of the list

# Note list is Mutable data type in python. We can add or remove elements from the list.

```

```python
# List Slicing in python !
# list_name[start_index:end_index:step] # note end_index is exclusive
#ex :

marks = [10, 20, 30, 40, 50]
print(marks[1:4])  # Output: [20, 30, 40]
print(marks[:3])   # Output: [10, 20, 30]
print(marks[2:5])  # Output: [30, 40, 50]
print(marks[2:])   #[30, 40, 50]
print(marks[::2])  # Output: [10, 30, 50] # [start:end:step]  # step is optional
```

```python
# Negative indexing in python list !

markes =[10, 20, 30, 40, 50]
print(markes[-1])  # Output: 50
print(markes[-2])  # Output: 40
print(markes[-3])  # Output: 30
print(markes[-4])  # Output: 20
print(marks[-3:-1])  # Output: [30, 40] # Note -1 is exclusive
```

```python
# List Methods in python !
#ex 
list1 = [1, 2, 3, 4, 5]
list1.append(6)  # Adding an element at the end of the list

print(list1)  # Output: [1, 2, 3, 4, 5, 6]

list2=[3,2,6,1,5,4]
list2.sort()  # Sorting the list in ascending order
print(list2)  # Output: [1, 2, 3, 4, 5, 6]

```

```python
# reverse() method in python list !
list3 = [1, 2, 3, 4, 5]
list3.sort(reverse=True)  # Sorting the list in descending order
print(list3)  # Output: [5, 4, 3, 2, 1]

```

```python
# for String values in list we can use sort() method to sort the list in ascending order and reverse() method to sort the list in descending order.
list = ["apple", "banana", "cherry", "date"]
list.sort()  # Sorting the list in ascending order
print(list)  # Output: ['apple', 'banana', 'cherry', 'date']
list.sort(reverse=True)  # Sorting the list in descending order
print(list)  # Output: ['date', 'cherry', 'banana', 'apple']

# Reverse() method in python list ! 
list2=['d','c','t','e']
list2.reverse()  # Reversing the list
print(list2)  # Output: ['s', 't', 'c', 'd']


```

```python
# insert() method in python list ! It is a little bit similar to append() method but the difference is that insert() method allows us to add an element at a specific index in the list, while append() method only adds an element at the end of the list.

list4 = [1, 2, 3, 4, 5]
list4.insert(2, 10)  # Adding an element at index 2
print(list4)  # Output: [1, 2, 10, 3, 4, 5]
```

```python
#remove() method in python list ! It is used to remove the first occurrence of a specified value from the list. If the value is not found, it raises a ValueError.
list5 = [1, 2, 3, 4, 5, 3]
list5.remove(3)  # Removing the first occurrence of 3
print(list5)  # Output: [1, 2, 4, 5, 3]
```

```python
#pop() method in python list ! It is used to remove an element from the list at a specified index and return the removed element. If the index is not specified, it removes and returns the last element of the list.
list6 = [1, 2, 3, 4, 5]
removed_element = list6.pop(2)  # Removing the element at index 2
print(removed_element)  # Output: 3
```
