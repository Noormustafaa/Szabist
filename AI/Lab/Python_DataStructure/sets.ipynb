```python
# sets in python is the collection of unique elements. It is unordered and unindexed. Sets are mutable, meaning you can add or remove elements from a set after its creation. However, the elements themselves must be immutable (e.g., numbers, strings, tuples).
#ex
set1 = {1, 2, 3, 4, 5}
print(set1)  # Output: {1, 2, 3, 4, 5}
print(type(set1))  # Output: <class 'set'>


set2={1,2,2,4,6,"Program","hello","Program"}
print(set2)  # Output: {1, 2, 4, 6, 'Program', 'hello'} (duplicates are removed)
print(type(set2))  # Output: <class 'set'>
print(len(set2))  # Output: 6 (number of unique elements in the set)
#print(set2[0])  # This will raise an error because sets are unordered and do not support indexing
print(set2)
```
