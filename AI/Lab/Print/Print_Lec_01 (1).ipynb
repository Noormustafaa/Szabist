```python
#Program 3: Custom Separator (sep parameter)
#Question: Print date components (Day, Month, Year) and join them using - using the sep parameter.

# Solution
day = 15
month = 8
year = 2026

print(day, month, year, sep="-")
print("AI", "Machine Learning", "Deep Learning", sep=" -> ")
```

```python
#Program 7: Legacy String Formatting (.format())
#Question: Format a string using positional and named placeholders with .format().

# Solution
model_name = "Neural Network"
accuracy = 95.8

# Positional formatting
print("Model: {}, Accuracy: {}%".format(model_name, accuracy))

# Named formatting
print("Algorithm: {algo} | Epochs: {ep}".format(algo="CNN", ep=50))




```
