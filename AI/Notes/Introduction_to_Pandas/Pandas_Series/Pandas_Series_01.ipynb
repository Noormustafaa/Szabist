```python
#pandas Series 
    # A Series is a one-dimensional labeled array capable of holding any data type (integers, strings, floating point numbers, Python objects, etc.). The axis labels are collectively referred to as the index. A Series is like a fixed-size dict in that you can get and set values by index label.
    #  It can also be thought of as a single column in a DataFrame.
    # Series Are vertical, one-dimensional arrays with an index. They can hold any data type, including objects. A Series is like a fixed-size dictionary in that you can get and set values by index label. It can also be thought of as a single column in a DataFrame.

import pandas as pd
import numpy as np

labels = ['a', 'b', 'c']
my_list = [10, 20, 30]
arr = np.array([10, 20, 30])
d={'a': 10, 'b': 20, 'c': 30}

# so lets create a series using these data structures

pd.Series(my_list) 
# output: 10
#         20
#         30
#         40 
# ARE IN VERTICAL SHAPE #and 0 ,1 ,2,3 are the index of the series



```

```python
# now we can also set the custom labels for the series using the index parameter
pd.Series(data=my_list, index=labels)
```

```python
# passing arrays 
pd.Series(arr) # note you cant pass multi dimensional arrays to series, it will throw an error
```

```python
# note we can also create series from a dictionary, the keys will be the index and the values will be the data
pd.Series(d) # no need to pass index parameter as the keys of the dictionary will be used as the index of the series
```
