```python
import numpy as np 
import pandas as pd
# Pandas Data frames 
# Collections of series, each series is a column in the data frame

# Creating a Data Frame 

# We can create a data frame by using list, arrays , dictionary 
```

```python
data = {
    'Name': ['John', 'Anna', 'Peter', 'Linda'],
    'Age': [28, 34, 29, 42],
    'City': ['New York', 'Paris', 'Berlin', 'London'],
    'Salary': [65000, 70000, 62000, 85000]
}


df =pd.DataFrame(data) # so this is how we can create a data frame from a dictionary, the keys will be the column names and the values will be the data in the columns.
print(df) # this will print the data frame in a tabular format, with the column names as the header and the index as the first column.
```

```python
# Creating a Data Frame from a list 
data_list = [
    ['John', 28, 'New York', 65000],
    ['Anna', 34, 'Paris', 70000],
    ['Peter', 29, 'Berlin', 62000],
    ['Linda', 42, 'London', 85000]
]
df1=pd.DataFrame(data_list)
print(df1)       # so in output we can see that the column names are not set,
         # so we can set the column names by passing a list of column names to the columns parameter of the DataFrame constructor.
# we can set this by passing a list of column names to the columns parameter of the DataFrame constructor.
column_names = ['Name', 'Age', 'City', 'Salary']
df1 = pd.DataFrame(data_list, columns=column_names) # so this is how we can set the columns manually 
print("")
print(df1)
```
