```python
import numpy as np 
import pandas as pd
# Pandas Data frames 
# Collections of series, each series is a column in the data frame

# Creating a Data Frame 

# We can create a data frame by using list, arrays , dictionary 
```

```python
data = {
    'Name': ['John', 'Anna', 'Peter', 'Linda'],
    'Age': [28, 34, 29, 42],
    'City': ['New York', 'Paris', 'Berlin', 'London'],
    'Salary': [65000, 70000, 62000, 85000]
}


df =pd.DataFrame(data) # so this is how we can create a data frame from a dictionary, the keys will be the column names and the values will be the data in the columns.
print(df) # this will print the data frame in a tabular format, with the column names as the header and the index as the first column.
```

```python
# Creating a Data Frame from a list 
data_list = [
    ['John', 28, 'New York', 65000],
    ['Anna', 34, 'Paris', 70000],
    ['Peter', 29, 'Berlin', 62000],
    ['Linda', 42, 'London', 85000]
]
df1=pd.DataFrame(data_list)
print(df1)       # so in output we can see that the column names are not set,
         # so we can set the column names by passing a list of column names to the columns parameter of the DataFrame constructor.
# we can set this by passing a list of column names to the columns parameter of the DataFrame constructor.
column_names = ['Name', 'Age', 'City', 'Salary']
df1 = pd.DataFrame(data_list, columns=column_names) # so this is how we can set the columns manually 
print("")
print(df1)
```

```python
# Pandas Selection and Indexing of Columns _03
df1

# What if we want to select a single column from the data frame, we can do that by using the column name as an attribute of the data frame.

#df1[0] # this will throw an error as the column name is not set, so we can set the column names by passing a list of column names to the columns parameter of the DataFrame constructor.

df1['Name'] # this will return a series of the Name column, we can also use the dot notation to access the column, but it is not recommended as it can lead to confusion with the methods of the data frame.
```

```python
# what if we want to select multiple columns from the data frame, we can do that by passing a list of column names to the DataFrame constructor.
# so we have to provide a list of list 
df1[['Name', 'Age']] # this will return a data frame with the selected columns, we can also use the dot notation to access the columns, but it is not recommended as it can lead to confusion with the methods of the data frame.

# So till now we have seen how to select a single column and multiple columns from the data frame, now what if we want to select a single row from the data frame, we can do that by using the iloc method of the data frame.
```

```python

#--- Creating a new column in the data frame 


df1['Country'] = ['USA', 'France', 'Germany', 'UK'] # so this is how we can create a new column in the data frame, we can also use the insert method of the data frame to insert a new column at a specific position in the data frame.
df1
```

```python
# So if you wan to remove a columns 
# we want to remove country column from the data frame, we can do that by using the drop method of the data frame.
#df1.drop("Country") # this will give error as we have to specify the axis parameter to drop a column, by default it is set to 0 which means we are dropping a row, so we have to set it to 1 to drop a column.
df1.drop("Country", axis=1) # this will drop the country column from the data frame, we can also use the inplace parameter to drop the column in place, so that we don't have to assign the result to a new variable.

# As we earlier discuss that axis 0 works on the x axis and axis 1 works on the y axis, so when we drop a column we are dropping it from the y axis, so we have to set the axis parameter to 1 to drop a column.
# till here country is not permenant dleted ex in next cell
```
