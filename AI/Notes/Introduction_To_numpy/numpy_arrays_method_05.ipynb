```python
# numpy arrays methods 
import numpy as np
arr1=np.array([[3, 2, 3, 4, 5], [6, 7, 8, 9, 10]])
np.min(arr1) # it will return the minimum value of the array
```

```python
arr2=np.array([[3, 2, 3, 4, 5], [6, 7, 8, 9, 33]])
np.max(arr2) # it will return the maximum value of the array
```

```python
arr3=np.array([[3, 2, 3, 4, 5], [6, 7, 8, 9, 33]])
print(arr3)
np.sum(arr3) # it will return the sum of all the elements in the array
```

```python
# now if we want the sum of the elements in each column we can use the axis parameter and set it to 0
# y-axis=0 means we want to sum the elements in each column
# x-axis=1 means we want to sum the elements in each row
sm=np.sum(arr3, axis=0) # it will return the sum of the elements in

print(sm)

smx = np.sum(arr3, axis=1) # it will return the sum of the elements in each row
print(smx)
```

```python
# now mean 

arr3.mean() # it will return the mean of all the elements in the array
```

```python
arr3.std() # it will return the standard deviation of all the elements in the array
```

```python
arr3.argmax() # it will return the index of the maximum value in the array
```

```python
arr3.argmin() # it will return the index of the minimum value in the array
```
