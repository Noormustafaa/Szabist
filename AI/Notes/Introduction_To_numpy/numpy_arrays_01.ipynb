```python
#Basic Array Creation 
import numpy as np

arr=np.array([1, 2, 3]) # we pass the list in array function and it returns a numpy array with the same elements.
arr
#print(type(arr)) # <class 'numpy.ndarray'>


# note in array we can store only one type of data, if we pass different types of data in array function it will convert all the data into one type.
# ex 
a=[1,2,3,4.6,"Noor"]
arr1=np.array(a)  # now we convert the list into array, it will convert all the data into string type.
arr1
# dtype=<u32 that means all the data is converted into string type.

```

```python
# Now What if we will not pass string 
#ex 
list=[1,2,4.5,6,7]
arr=np.array(list)
arr  
# this will convert all the data into float type because we have one float value in the list.



```

```python
# note in arrays and list there are two concepts 
#1 Vectors And matrix 

#Vectors Are somtheing like a list, it is a one dimensional array.
#ex [2,3,4,5,6] this is a vector because it is one dimensional array.

# matrix is a two dimensional array, it is something like a table with rows and columns.
# ex [[1,2,3],[4,5,6]] this is a matrix because it is two dimensional array. But note list 
#does not support two dimensional array, it is only one dimensional array. But we can create a list of lists to create a two dimensional array.
#ex 
l=[[1,2,3],[4,5,6]]
l   # it's output is in one dimensoional .



```

```python
# but when we convert it into array it will become two dimensional array.
arr=np.array(l)
arr

# if there are more than two dimensional array it is called tensors.(like Tensorflow library )
```
