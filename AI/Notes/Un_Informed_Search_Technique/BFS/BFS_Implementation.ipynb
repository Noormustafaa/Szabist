```python
from collections import deque

# The image tree graph is defined as a dictionary (Adjacency List).

tree_graph = {
    'A': ['B', 'C', 'D'],
    'B': ['E', 'F'],
    'C': ['G', 'H'],
    'D': ['I'],
    'E': ['J', 'K'],
    'F': [],
    'G': ['L'],
    'H': [],
    'I': ['M'],
    'J': [],
    'K': ['N'],
    'L': [],
    'M': [],
    'N': []
}

# 2. Breadth-First Search Function
def bfs_tree(graph, start_node):
    visited = []                  # To store the visited nodes
    queue = deque([start_node])   # Queue from starting node 'A'
    
    while queue:
        current_node = queue.popleft() # We deque the element from left side of Queue

  
        
        if current_node not in visited:
            visited.append(current_node) # After Visiting nodes we add in visited

            

            #print('hello now iam visited value ',visited)
            
           
            for neighbor in graph[current_node]: # graph['A']--dictionary_name[key]  # it returns the value of key from the dic

                #print("hello Iam neighbour",neighbor)

                queue.append(neighbor)
                
    return visited

# --- Execution ---
print("BFS Traversal Order for the given tree:")
traversal_result = bfs_tree(tree_graph, 'A')
print(traversal_result)
```
