```python
# Random Generation Function !
# Random Library is used to generate random numbers in python.
# and numpy allows us to create arrays of random numbers with various distributions.

import numpy as np
random1 =np.random.rand(10) # it will generate 5 random numbers between 0 and 1
print(random1)
```

```python
# For Generating Random Integers, we can use the randint() function from the numpy library. This function allows us to specify a range and the number of random integers we want to generate.
rand2=np.random.randint(5)
print(rand2)

# Now for generating random numbers from a specific range 
rand3=np.random.randint(1,10) # it will generate 5 random numbers between 1 and 10
print(rand3)

rand4=np.random.randint(1,10,5) # it will generate 5 random numbers between 1 and 10
print(rand4) # it will generate those numbers in array 
```
