```python

import numpy as np  
import pandas as pd 
import matplotlib.pyplot as plt  

x=np.linspace(0,5,11) # (starting index, inding, steps) 
x

print(x)  # 11 elements in between 0 and 5 

y=x**3  # cube of these values 

y

print(y) # so these are all the values of y

plt.plot(x,y)  # plt.plot here plot create the line graph , So want to plot these points 

# or we can also write (x=x and y=y) 

# now we want to add labels to them ! 
plt.title("This is Hello Graph ")

plt.xlabel("Iam X-axes ")

plt.ylabel("Iam Y-axes ")  # not included ! 

# -- not included fig=plt.figure() # now it will now showing any thing but 

# -- not included fig.add_axes([0.1,0.1,1,0.8])#in this we will add here 4 things 

```

```python
# Subplots in matplotlib 

# note canvas is an empty space where inside it we can create plots . 
# like when we write plt.plot so on canvas plots were created 
# so in plt.subplot() method we need to provide number of rows and number of columns 

# if you will write plt.subplot(2,2) # this will create inside the canvas two rows and two columns 

# and When you write plt.subplot(2,2,1) that means you are selecting firt portion of it 

plt.subplot(2,2,1)  # first plot 
plt.plot(x,y)
plt.subplot(2,2,4)

plt.plot(x,y) # if you want to specify plot on any canva  # in 4rth one 


plt.subplot(2,2,3)

plt.plot(y,x)


plt.subplot(2,2,2) # now we have selected the 4rth plot    # note but still we have not plot anything on it !

plt.plot(x**2,x**2)  # so this is how we can plot 
```
